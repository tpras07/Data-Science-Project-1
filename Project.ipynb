There were a lot of issues in the data that we have cleaned using the code written above. We will enumerate them below along with a short description of the fixes we applied:  
1. Eight MMSE scores were above the maximum of 30. Since we can't guess their correct values, we changed the corresponding scores to missing.
2. The hippocampal volume contained extremely small values, which suggests that they were mistakenly recorded in cubic centimeters rather than in cubic millimeters. We converted them by multiplying by 1000.
3. Missing data: After cleaning, we found 106 instances of missing values in BMI (10%), 170 in MOCA (8%), 212 in hippocampal volume (212), and 8 in MMSE. We retained the rows containing the missing values instead of replacing them because they contained valuable information.
4. Initially, the diagnosis column contained nine distinct labels; there were a mismatch of capitalization and abbreviations/terminologies. We standardized them into four categories: Normal, Mild Cognitive Impairment, Dementia, and Alzheimer's Disease.
   
We found no duplicate rows. We also checked for possible duplicable combinations of ID and visit number: we found none. 

```python
data["risk_count"] = data["hypertension"].map({"Yes":1, "No":0}).astype(int) + data["apoe4_carrier"].map({"Yes":1, "No":0}).astype(int)
```

```python
data.head()
```

We created a new column, risk_count, by combining hypertension and apoe4_carrier columns. We converted 'Yes' to 1 and 'No' to 0, and then we summed them to get the risk_count. It might help us to find out how hypertension and apoe4_carrier are associated with cognitive performance.

```python
risky_patients = data[data["risk_count"]>=1]
risky_patients.head(10)
```

```python
old_patients = data[
    (data["age"] >= 75) &
    (data["moca"].notna())
]

display(old_patients.head(10))
print("Number of records:", len(old_patients))
```

We applied two filters to get two different subsets of the dataset.
1. Filter 1: The first filter checks for the presence of risk indicators by selecting observations with risk_count >= 1. It's interesting because it could give us an insight connecting risk indicators with cognitive performance.
2. Filter 2: The second filter selects older participants with available MOCA scores by selecting observations with age >= 75 with non-missing MOCA scores, giving us 1030 records. It's interesting because it gives us the cognitive performance of older patients. 

```python
categorical_cols = [
    "sex",
    "hypertension",
    "apoe4_carrier",
    "diagnosis"
]

for col in categorical_cols:
    print(f"\n{col}:")
    print(data[col].unique())
```

The dataset contains four categorical variables: sex, hypertension, apoe4 carrier status, and diagnosis. Sex, hypertension, and apoe4 status each contain 2 categories, while diagnosis contains 4 categories following standardization. 
Among these variables, diagnosis is the most important for grouping because it allows us to compare cognitive tests across different classifications. Hyptertension and apoe4 status are also important because they can allow us to form links between indicators and cognitive performance.

```python
data.groupby("diagnosis", observed=True)["mmse"].mean()
```

```python
risk_summary = data.groupby("risk_count").agg(
    mean_mmse=("mmse", "mean"),
    mean_moca=("moca", "mean"),
    mean_age=("age", "mean"),
    count=("mmse", "count")
)

display(risk_summary.round(2))
```

We did two groupings:
1. Grouping by Diagnosis:
We grouped the dataset by diagnosis and calculated the average MMSE score for each category. The results show that observations classified as Normal have the highest average MMSE score (28.18), followed by Mild Cognitive Impairment (24.51), Dementia (18.71), and Alzheimer's Disease (18.50). We can see the direct correlation between diagnostic classification and cognitive performance.

2. Grouping by Risk Count:
We also grouped the dataset by risk_count and calculated the average MMSE, MoCA, and age for each group. Both cognitive test scores decrease as the number of recorded risk indicators increases. The mean MMSE decreases from 26.94 in the group with no risk indicators to 24.95 in the group with both indicators. Similarly, the mean MoCA decreases from 24.37 to 22.43.

```python
sorted_data = data.sort_values(["participant_id", "visit_date"])

change = sorted_data.groupby("participant_id").agg(
    first_mmse=("mmse", "first"),
    last_mmse=("mmse", "last"),
    visits=("mmse", "size")
)

change = change[change["visits"] >= 2]

change["mmse_change"] = change["last_mmse"] - change["first_mmse"]

print(change["mmse_change"].describe())
```
